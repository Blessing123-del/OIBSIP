## Missing Data Handling Strategy

Missing values were assessed based on the proportion of missing data and the importance of each variable.

- Age: Missing age values will be replaced with the median age because age is a numerical variable and the median is less affected by extreme values than the mean.
- Embarked: The two missing values will be replaced with the mode because Embarked is a categorical variable and the mode represents the most common category.
- Cabin: Because a very large proportion of Cabin values are missing, direct statistical imputation would not be appropriate. The missing values will be represented as "Unknown" to preserve the records while clearly indicating that the cabin information was unavailable.

```python
age_median = df["Age"].median()

df["Age"] = df["Age"].fillna(age_median)

print("Age median used for imputation:", age_median)
print("Remaining missing Age values:", df["Age"].isnull().sum())
```

```python
embarked_mode = df["Embarked"].mode()[0]

df["Embarked"] = df["Embarked"].fillna(embarked_mode)

print("Embarked mode used for imputation:", embarked_mode)
print("Remaining missing Embarked values:", df["Embarked"].isnull().sum())
```

```python
df["Cabin"] = df["Cabin"].fillna("Unknown")

print("Remaining missing Cabin values:", df["Cabin"].isnull().sum())
```

## Data Type Correction

Data types were reviewed and corrected to ensure that each variable is stored in an appropriate format. Passenger IDs and text-based variables were stored as strings, while numerical variables such as Age and Fare were assigned numerical data types. Correct data types improve memory usage, calculations, filtering, and downstream analysis.

```python
numeric_columns = df.select_dtypes(include=np.number).columns

print("Numerical columns:")
print(list(numeric_columns))
```

```python
outlier_report = []

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    
    outlier_report.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers)
    })

outlier_report = pd.DataFrame(outlier_report)

outlier_report
```

## Outlier Detection

The Interquartile Range (IQR) method was used to identify potential outliers in numerical variables. The IQR was calculated as the difference between the third quartile (Q3) and first quartile (Q1). Values below Q1 − 1.5 × IQR or above Q3 + 1.5 × IQR were flagged as potential outliers.

Outliers were reviewed before deciding whether to remove, cap, or retain them. Genuine extreme values were retained because removing valid observations could result in loss of useful information.

```python
for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]
    
    print(f"\n{column}: {len(outliers)} outliers")
    
    if len(outliers) > 0:
        print(outliers[column].head(10).to_list())
```
