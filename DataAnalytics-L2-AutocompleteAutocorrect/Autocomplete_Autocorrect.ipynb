```python
with open("alice_in_wonderland.txt", "r", encoding="latin-1") as file:
    text = file.read()

print(text[:1000])
```

```python
import re

# Remove extra spaces and blank lines
clean_text = re.sub(r'\s+', ' ', text).strip()

print(clean_text[:1000])
```

```python
words = re.findall(r"\b[a-zA-Z]+\b", clean_text.lower())

print(words[:50])
print("Total words:", len(words))
```

```python
from collections import Counter

word_counts = Counter(words)

print(word_counts.most_common(20))
```

```python
from nltk.util import ngrams

bigrams = list(ngrams(words, 2))

print(bigrams[:20])
print("Total bigrams:", len(bigrams))
```

```python
from collections import defaultdict, Counter

next_words = defaultdict(Counter)

for word1, word2 in bigrams:
    next_words[word1][word2] += 1

print(next_words["alice"].most_common(5))
```

```python
def autocomplete(word, n=3):
    word = word.lower()
    
    if word in next_words:
        return [w for w, count in next_words[word].most_common(n)]
    else:
        return []

print(autocomplete("alice"))
```

```python
test_words = ["alice", "the", "queen", "rabbit", "said"]

for word in test_words:
    print(word, "→", autocomplete(word))
```

```python
from spellchecker import SpellChecker

spell = SpellChecker()

print(spell.correction("alicce"))
```

```python
def autocorrect(word):
    word = word.lower()
    correction = spell.correction(word)
    return correction

print(autocorrect("alicce"))
print(autocorrect("wondrland"))
print(autocorrect("rabbitt"))
```

```python
test_words = [
    "alicce",
    "wondrland",
    "rabbitt",
    "queenn",
    "beggining",
    "curiouse",
    "adventurre",
    "gryphon",
    "chesshire",
    "madde"
]

for word in test_words:
    print(word, "→", autocorrect(word))
```

```python
correct_words = [
    "alice",
    "wonderland",
    "rabbit",
    "queen",
    "beginning",
    "curious",
    "adventure",
    "gryphon",
    "cheshire",
    "mad"
]

predicted_words = [
    "alice",
    "wonderland",
    "rabbit",
    "queen",
    "beginning",
    "curious",
    "adventure",
    "gryphon",
    "cheshire",
    "made"
]

correct_predictions = sum(
    actual == predicted
    for actual, predicted in zip(correct_words, predicted_words)
)

accuracy = correct_predictions / len(correct_words)

print("Correct predictions:", correct_predictions)
print("Total tests:", len(correct_words))
print(f"Autocorrect Accuracy: {accuracy:.2%}")
```

```python
test_pairs = [
    ("alice", "was"),
    ("white", "rabbit"),
    ("march", "hare"),
    ("mad", "hatter"),
    ("queen", "of"),
    ("king", "of"),
    ("said", "alice"),
    ("little", "girl"),
    ("long", "time"),
    ("beginning", "to")
]

correct_predictions = 0

for word, expected in test_pairs:
    predictions = autocomplete(word)
    
    if expected in predictions:
        correct_predictions += 1
    
    print(f"{word} → {predictions} | Expected: {expected}")

accuracy = correct_predictions / len(test_pairs)

print(f"\nAutocomplete Top-3 Accuracy: {accuracy:.2%}")
```
